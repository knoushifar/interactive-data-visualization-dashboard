## Loading data

```python
import pandas as pd
import numpy as np
import plotly.express as px
import plotly.graph_objects as go
from plotly.subplots import make_subplots

irisdf = px.data.iris()
irisdf.head(5)
```

```python
duplicates = irisdf[irisdf.duplicated()]
print("Number of duplicates:", len(duplicates))
```

```python
irisdf = irisdf.drop_duplicates()
```

```python
irisdf.describe()
```

```python
species=irisdf["species"].unique()
print(species)
```

## Data visualization

## Scatter plot 

```python
fig = px.scatter(
    irisdf,
    x="petal_length",
    y="petal_width",
    color="species",
    facet_col='species',
    hover_data={
        'species': False,
        'sepal_length': ':.2f',
        'petal_width': True,
        'sepal_width': ':.2f',
    }
)
fig.show()
fig.write_html("subplot.html")
```

```python
fig = px.scatter(
    irisdf,
    x="petal_length",
    y="petal_width",
    color="species",
    hover_data={
        'species': False,
        'sepal_length': ':.2f',
        'petal_width': True,
        'sepal_width': ':.2f',
    }
)

cluster0 = [dict(type="circle",
                 xref="x", yref="y",
                 x0=min(irisdf[irisdf['species'] == "setosa"]["petal_length"]),
                 y0=min(irisdf[irisdf['species'] == "setosa"]["petal_width"]),
                 x1=max(irisdf[irisdf['species'] == "setosa"]["petal_length"]),
                 y1=max(irisdf[irisdf['species'] == "setosa"]["petal_width"]),
                 line=dict(color="Blue"))]

cluster1 = [dict(type="circle",
                 xref="x", yref="y",
                 x0=min(irisdf[irisdf['species'] == "versicolor"]["petal_length"]),
                 y0=min(irisdf[irisdf['species'] == "versicolor"]["petal_width"]),
                 x1=max(irisdf[irisdf['species'] == "versicolor"]["petal_length"]),
                 y1=max(irisdf[irisdf['species'] == "versicolor"]["petal_width"]),
                 line=dict(color="Red"))]

cluster2 = [dict(type="circle",
                 xref="x", yref="y",
                 x0=min(irisdf[irisdf['species'] == "virginica"]["petal_length"]),
                 y0=min(irisdf[irisdf['species'] == "virginica"]["petal_width"]),
                 x1=max(irisdf[irisdf['species'] == "virginica"]["petal_length"]),
                 y1=max(irisdf[irisdf['species'] == "virginica"]["petal_width"]),
                 line=dict(color="Green"))]

fig.update_layout(
    updatemenus=[
        dict(buttons=list([
            dict(label="None",
                 method="relayout",
                 args=["shapes", []]),
            dict(label="Cluster 0 (Setosa)",
                 method="relayout",
                 args=["shapes", cluster0]),
            dict(label="Cluster 1 (Versicolor)",
                 method="relayout",
                 args=["shapes", cluster1]),
            dict(label="Cluster 2 (Virginica)",
                 method="relayout",
                 args=["shapes", cluster2]),
            dict(label="All Clusters",
                 method="relayout",
                 args=["shapes", cluster0 + cluster1 + cluster2])
        ]),
        )
    ]
)

fig.show()
fig.write_html("clustered.html")
```

```python
fig = px.scatter(
    irisdf,
    x="petal_length",
    y="petal_width",
)

fig.update_layout(
    updatemenus=[
        dict(
            buttons=list([
                dict(label="Petal Dimensions",
                     method="update",
                     args=[{"x": [irisdf["petal_length"]],
                            "y": [irisdf["petal_width"]]},
                           {"xaxis": {"title": "Petal Length"},
                            "yaxis": {"title": "Petal Width"}}]),
                dict(label="Sepal Dimensions",
                     method="update",
                     args=[{"x": [irisdf["sepal_length"]],
                            "y": [irisdf["sepal_width"]]},
                           {"xaxis": {"title": "Sepal Length"},
                            "yaxis": {"title": "Sepal Width"}}])
            ]),

        )
    ],
    title="Iris Dataset: Toggle Between Petal and Sepal Dimensions",
    xaxis_title="Petal Length",
    yaxis_title="Petal Width"
)
fig.show()
fig.write_html("dropdown.html")
```

```python
fig = px.scatter(
    irisdf,
    x="petal_length",
    y="petal_width",
    animation_frame="species", 
    animation_group="petal_length"
)

fig.show()

```

## Bar plot

```python
fig = make_subplots(rows=1, cols=len(species), subplot_titles=species)

for i, sp in enumerate(species, start=1):
    species_data = irisdf[irisdf['species'] == sp]
    count = species_data.shape[0]

    fig.add_trace(
        go.Bar(x=[sp], y=[count], name=sp),
        row=1, col=i
    )

fig.update_layout(
    title_text='Count of Different Species',  
    xaxis_title='Species',
    yaxis_title='Count',
)
fig.show()

fig.write_html("barplot.html")
```

## Box plot

```python
fig = px.box(irisdf, x="species", y="sepal_length")
fig.show()
fig.write_html("boxplot.html")
```

## Histogram of petal lenght


```python
fig = px.histogram(irisdf, x="petal_length")
fig.show()
fig.write_html("histogramplot.html")
```

## Normalization

```python
import scipy
data_norm_to_0_1 = [
    number/scipy.linalg.norm(irisdf["petal_length"]) for number in irisdf["petal_length"]]
fig = px.histogram(data_norm_to_0_1)
fig.show()
fig.write_html("normalhistogramplot.html")
```
